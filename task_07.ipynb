```python
#Problem 1: Smart Light Controller
class Smartlight:
  def __init__(self, name):
    self.name = name
    self.status = "OFF"

  def turn_on(self):
    self.status = "ON"
    print(f"{self.name} is {self.status}")

  def turn_off(self):
    self.status = "OFF"
    print(f"{self.name} is {self.status}")

smartlight1 = Smartlight("Bedroom light")
# smartlight1.turn_on()
# smartlight1.turn_off()
action = input("Enter the choice(ON/OFF) : ")

if action == "ON":
  smartlight1.turn_on()
elif action == "OFF":
  smartlight1.turn_off()
else:
  print("Invalid choice")
```

```python
#Problem 2: Employee ID Card System
class Employee:
  def __init__(self, emp_name, emp_id, emp_dept):
    self.emp_name = emp_name
    self.emp_id = emp_id
    self.emp_dept = emp_dept

  def display_card(self):
    print("Employee ID Card")
    print(f"Name: {self.emp_name}")
    print(f"ID: {self.emp_id}")
    print(f"Department: {self.emp_dept}")

emp1 = Employee("Rahul", "EMP102", "IT")
emp1.display_card()
```

```python
#Problem 3: Mobile Contact Record
class MobileContact:
  def __init__(self, name, phone_number, email):
    self.name = name
    self.phone_number = phone_number
    self.email = email

  def display_contact(self):
    print("Contact Details")
    print(f"Name: {self.name}")
    print(f"Phone Number: {self.phone_number}")
    print(f"Email: {self.email}")

contact1 = MobileContact("Anitha", "9876543210","anitha@gmail.com")
contact1.display_contact()
```

```python
#Problem 4: Product Price Tag Generator
class Product:
  def __init__(self, name, price):
    self.name = name
    self.price = price


  def generate_tag(self):
    print("Product Price Tag")
    print(f"Name: {self.name}")
    print(f"Price: {self.price}")

product1 = Product("Headphones", 2499)
product1.generate_tag()
```

```python
#Problem 5: Movie Rating Display System
class Movie:
  def __init__(self, title, rating):
    self.title = title
    self.rating = rating

  def display_rating(self):
    print("Movie Rating")
    print(f"Title: {self.title}")
    print(f"Rating: {self.rating}")

movie1 = Movie("Inceptiom", 4.8)
movie1.display_rating()
```

```python
#Problem 6: Delivery Address Manager
class DeliveryAddress:
  def __init__(self, name, address):
    self.name = name
    self.address = address

  def display_address(self):
    print("Delivery Details")
    print(f"Name: {self.name}")
    print(f"Address: {self.address}")

samp = DeliveryAddress("Suman", "Hyderabad")
DeliveryAddress.display_address(samp)
```
