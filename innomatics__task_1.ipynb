```python
#2. Pass / Fail Analyzer
marks = [45, 78, 90, 33, 60]
a = len(marks)
pass_count = 0
fail_count = 0

for mark in marks:
  if mark >= 50:
    pass_count += 1
  else:
    fail_count += 1

print(f"Total students are {a}")
print(f"Total pass students are {pass_count}")
print(f"Total fail students are {fail_count}")
```
