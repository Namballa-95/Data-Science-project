```python
#Problem 2: Inventory Reorder Alert System
inventory = {
    "Rice": 20,
    "Sugar": 10,
    "Oil": 5,
    "Wheat": 18
}

def reorder_alert(inventory):
  for item, quantity in inventory.items():
    if quantity < 15:
       print(f"Reorder alert: {item} is running low. Quantity: {quantity}")
    else:
      print(f"{item} is in stock. Quantity: {quantity}")

reorder_alert(inventory)
```

```python
#Problem 3: Student Result Processing System
def process_result(marks):
    total = 0

    # Calculate total using loop
    for mark in marks:
        total += mark

    average = total / len(marks)

    # Determine result
    if average >= 50:
        result = "Pass"
    else:
        result = "Fail"

    print(f"Average Marks: {average}")
    print(f"Result: {result}")

    return result



student_marks = [60, 45, 70, 55]
process_result(student_marks)
```

```python
#Problem 4: Cab Fare Estimator with Retry Option
def calculate_fare(distance, is_peak):
    base_fare = 50
    per_km_rate = 12

    # Calculate normal fare
    fare = base_fare + (distance * per_km_rate)

    # Add peak hour charge (25%)
    if is_peak:
        fare += fare * 0.25

    return fare


# Retry system using while loop
while True:
    try:
        distance = float(input("Enter distance in km: "))
        peak_input = input("Is it peak hour? (yes/no): ").lower()

        if peak_input not in ["yes", "no"]:
            print("Please enter 'yes' or 'no'.\n")
            continue

        is_peak = True if peak_input == "yes" else False

        total_fare = calculate_fare(distance, is_peak)

        print(f"Total Fare: ₹{total_fare:.2f}")
        break

    except ValueError:
        print("Invalid input. Please enter numeric distance.\n")
```

```python
#Problem 5: Employee Attendance Eligibility Checker
def check_attendance(attendance_list):
    present_count = 0

    # Count present days using loop
    for status in attendance_list:
        if status == "P":
            present_count += 1

    total_days = len(attendance_list)

    # Avoid division by zero
    if total_days == 0:
        return "No Attendance Data"

    percentage = (present_count / total_days) * 100

    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"


# Example usage
attendance = ["P", "A", "P", "P", "A", "P", "P", "P"]
result = check_attendance(attendance)

print("Attendance Result:", result)
```

```python
#Problem 6: Password Strength Checker
def check_password_strength(password):
    has_digit = False
    has_special = False

    # Rule 1: Minimum length check
    if len(password) < 8:
        return "Weak Password (Minimum length is 8)"

    # Loop through characters
    for char in password:
        if char.isdigit():
            has_digit = True
        if char in "@#$":
            has_special = True

    # Final validation
    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password (Must include at least one digit and one special character @#$)"


# Example usage
pwd = input("Enter password: ")
result = check_password_strength(pwd)
print(result)
```
